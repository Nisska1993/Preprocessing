# Encoding

Transformer les valeurs non numeriques en valeurs numeriques.

L'encodage ordinal permet d'encoder les variables ordinales (dont les valeurs forment une hiérarchie).
Encodage OneHot pour les variables (quand il n'y a pas d'ordre).

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import OrdinalEncoder
import sklearn
print(sklearn.__version__)
```

# Load Diamonds

```python
df = sns.load_dataset("diamonds")
df.head()
```

# Utilisation de Ordinal Encoder 
* Réservé a l'encodage des variables Ordinales (avec heirarchie).
* L'argument categorie permet de gérer l'ordre
* handle_unknown ="use_encoded_value" unkonwn_value =- permet de rendre le code plus robuste

```python
clear_order = ['Fair', 'Good', 'Very Good', 'Premium', 'Ideal']
color_order = ['J', 'I', 'H', 'G', 'F', 'E', 'D']
clarity_order = ['I1', 'SI2', 'SI1', 'VS2', 'VS1', 'VVS2', 'VVS1', 'IF']

encoder = OrdinalEncoder(categories=[clear_order, color_order, clarity_order], handle_unknown='use_encoded_value', unknown_value=-1)
encoder.fit(df[['cut', 'color', 'clarity']])
X = encoder.transform(df[['cut', 'color', 'clarity']])
```
